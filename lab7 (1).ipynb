## 1. A quick reminder

Object-oriented programming (OOP) groups related data and behavior in objects.
A class describes how to create them. Each bank account has its own owner and
balance, and all accounts use the same methods. This helps when you have many
similar things that follow the same rules: you can reuse the account setup and
transaction code whenever you create another account or make another transfer.

Encapsulation keeps an object's data with the methods that manage it, so those
methods can check the rules before changing anything. Abstraction gives the
caller an operation such as `transfer_to()` that handles the checks and updates
for them. We'll use it to replace repeated `if`/`else` blocks with method calls.
Ordinary functions can remove repetition too; a class keeps that reusable
behavior with the data it works on.

## 4. Write one transfer, then repeat it

Alice wants to send Bob Rs 250. Their objects hold the account data, so we can
read and change their balances. So far, though, the program using those objects
has to carry out every step of the transfer itself.

As you read the worked example, find where it does each of these jobs:

1. Rejects an amount that is zero or negative.
2. Rejects a transfer to the same account. The expression `alice is bob` checks
   whether both names refer to the same object.
3. Rejects an amount greater than Alice's balance.
4. Subtracts from Alice, adds to Bob, and appends an entry to each history when
   all the checks pass. Each history tuple holds the direction, the other
   owner's name, and the amount.

All four updates belong in the `else` branch. If a transfer is rejected, both
accounts must stay unchanged. The setup resets the balances, so Alice's extra
Rs 100 from the previous example does not carry over. Each setup line assigns
the owner, balance, and history together, using the attributes you just practised.

Work out the two balances before running the example.

````python
alice = AccountRecord()
alice.owner, alice.balance, alice.history = "Alice", 1200, []
bob = AccountRecord()
bob.owner, bob.balance, bob.history = "Bob", 300, []

amount = 250
if amount <= 0:
    print("Transfer rejected: amount must be positive.")
elif alice is bob:
    print("Transfer rejected: choose another account.")
elif amount > alice.balance:
    print("Transfer rejected: insufficient balance.")
else:
    alice.balance -= amount
    bob.balance += amount
    alice.history.append(("sent", bob.owner, amount))
    bob.history.append(("received", alice.owner, amount))
    print("Transfer complete.")

print("Balances:", alice.balance, bob.balance)
print("Alice's history:", alice.history)
````

## 5. Give the account reusable methods

Start with the repeated account setup. Read the class below:

- `def __init__(self, owner, opening_balance):` defines the setup Python runs
  when you create an account. It stores the supplied name and amount on that object.
- `self` is the account being set up or used. Its balance is `self.balance`.
- `def get_balance(self):` defines a method you can call on an account.
  The line `return self.balance` sends its balance back to the caller.

Run the example and compare `BankAccount("Alice", 1200)` with the separate
creation and attribute-assignment lines you used earlier.

````python
class BankAccount:
    def __init__(self, owner, opening_balance):
        self.owner = owner
        self.balance = opening_balance

    def get_balance(self):
        return self.balance

alice = BankAccount("Alice", 1200)
bob = BankAccount("Bob", 300)
current_balance = alice.get_balance()
print("Alice:", current_balance)
print("Bob:", bob.get_balance())
````

> Expected output:
> 
> ```text
> Alice: 1200
> Bob: 300
> ```

When you call `alice.get_balance()`, Python passes Alice's object as `self`
for you. The example stores the returned number in `current_balance` and prints
it. You can use a returned number in a calculation too.

## 6. Move the transfer checks into the account

You can now create an account in one line. Add a method called
`transfer_to(self, recipient, amount)` to handle the transfer procedure you
repeated earlier. The improved class uses `_balance` and `_history` for internal
data, with methods to read or update it. A leading underscore asks callers to
use the methods as these are known as protected attributes and should not be accessed 
directly; Python still allows direct access to those attributes.

### Guided practice: Complete the transfer method

Fill the five blanks in the class below:

1. In `get_history()`, use the list method that returns a separate copy.
2. Choose the amount boundary that rejects zero and negative values.
3. Compare the amount with the account's internal balance attribute.
4. Deduct the transfer amount from the sender.
5. Add that amount to the recipient.

Keep your inline blocks handy. You're using the same checks and updates, this
time inside one method. The line `raise ValueError(...)` stops account creation
with an error if the opening balance is negative. Our examples use valid balances.

Fill all the blanks, then run the class definition. This replaces the earlier
`BankAccount` definition. Create fresh accounts using the example that follows;
objects you created earlier keep their previous implementation.

````python
class BankAccount:
    def __init__(self, owner, opening_balance):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative.")
        self.owner = owner
        self._balance = opening_balance
        self._history = []

    def get_balance(self):
        return self._balance

    def get_history(self):
        return self._history.copy()  # return a copy of the list

    def transfer_to(self, recipient, amount):
        if amount <= 0:  # reject zero and negative amounts
            return "Transfer rejected: amount must be positive."
        if recipient is self:
            return "Transfer rejected: choose another account."
        if amount > self._balance:  # compare with the internal balance
            return "Transfer rejected: insufficient balance."

        self._balance -= amount  # deduct the transfer amount
        recipient._balance += amount  # credit the same amount
        self._history.append(("sent", recipient.owner, amount))
        recipient._history.append(("received", self.owner, amount))
        return f"Transferred Rs {amount} to {recipient.owner}."
````

## 8. Your turn: Build the café's coffee machine

Meera wants café staff to make coffee with one call. Write the steps that make
that possible, using your bank transfer as a model. Complete the single
`CoffeeMachine` class in the starter cell. Each coffee uses `100` ml of water.

Work through the methods in this order:

1. In `__init__(self, water_ml)`, reject a negative starting amount with
   `ValueError`, as the bank constructor does. Otherwise, store the starting
   water in `self._water_ml`.
2. Make `get_water(self)` return `self._water_ml`.
3. In `refill_water(self, amount)`, return `False` immediately if the amount
   is zero or negative, leaving the water unchanged. Otherwise, add the amount
   and return `True`.
4. In `make_coffee(self)`, check the water first. If it is below `100`, return
   `"Not enough water."` without changing the water or printing preparation
   messages. Otherwise, print `"Heating water..."`, `"Brewing coffee..."`, and
   `"Pouring coffee..."` in that order. Deduct `100` ml and return `"Coffee ready!"`.

Replace each `pass` with your code. Keep water checks and updates inside the
class so that staff can use its methods without writing their own conditions.
As with the bank, reject an invalid request before changing any data. The
preparation messages simulate the steps behind `make_coffee()`.

Run your completed class and then the usage cell below it. Before running the
usage cell, calculate the water remaining: start at `150`, use `100`, reject
another coffee, add `200`, and use `100` again. Compare your output with the
example, then run the extra checks.

````python
class CoffeeMachine:
    def __init__(self, water_ml):
        # Reject negative starting water; otherwise store the internal water level.
        if water_ml < 0:
            raise ValueError("Water level cannot be negative.")
        self._water_ml = water_ml

    def get_water(self):
        # Return the current water level.
        return self._water_ml

    def refill_water(self, amount):
        # Check the amount, then refill. Return True or False.
        if amount <= 0:
            return False
        self._water_ml += amount
        return True

    def make_coffee(self):
        # Check water, simulate the three steps, use 100 ml, and return a message.
        if self._water_ml < 100:
            return "Not enough water."
        print("Heating water...")
        print("Brewing coffee...")
        print("Pouring coffee...")
        self._water_ml -= 100
        return "Coffee ready!"
````

````python
machine = CoffeeMachine(150)
print(machine.make_coffee())
print("Water:", machine.get_water())
print(machine.make_coffee())
print("Refilled:", machine.refill_water(200))
print(machine.make_coffee())
print("Water:", machine.get_water())
````

> Expected output:

> Heating water...
> Brewing coffee...
> Pouring coffee...
> Coffee ready!
> Water: 50
> Not enough water.
> Refilled: True
> Heating water...
> Brewing coffee...
> Pouring coffee...
> Coffee ready!
> Water: 150
> ```

One call to `make_coffee()` checks the water, runs the preparation steps, and
