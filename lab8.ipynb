The King has decided to organise his royal court. Knights, princes, royal guards and other members of the kingdom all have different responsibilities, but they also share certain characteristics.

## Part 1: Meet the Members of the Royal Court

Before building the royal court, let's understand how inheritance works.

A *Prince* and a *Knight* may have different responsibilities, but both belong to the kingdom. They might both have a name, age and kingdom.

Instead of writing these attributes separately in both classes, we can create a common parent class called `RoyalMember`.

````python
class RoyalMember:
    def __init__(self, name, age, kingdom):
        self.name = name
        self.age = age
        self.kingdom = kingdom

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}.")

    def give_advice(self):
        print(f"{self.name} has no official recommendation for the kingdom.")
````

### Task 1: Create your first royal member

Print Arthur's name, age and kingdom using the object's attributes.

Call the `introduce()` method.

Expected introduction:

`I am Arthur of Eldoria.`

````python
arthur = RoyalMember("Arthur", 25, "Eldoria")
print(f"Name: {arthur.name}")
print(f"Age: {arthur.age}")
print(f"Kingdom: {arthur.kingdom}")
arthur.introduce()
````

## Part 2: The Knights Join the Kingdom (Inheritance)

The King now wants to recruit knights to protect the kingdom.
Every knight has a name, age and kingdom, just like other royal members. However, knights also have additional attributes, such as their weapon and number of battles won.

Instead of rewriting all the attributes, we can make the `Knight` class inherit from `RoyalMember`.

````python
class Knight(RoyalMember):
    def __init__(self, name, age, kingdom, weapon, battles_won):
        super().__init__(name, age, kingdom)
        self.weapon = weapon
        self.battles_won = battles_won

    def display_details(self):
        print(f"Knight: {self.name}")
        print(f"Kingdom: {self.kingdom}")
        print(f"Weapon: {self.weapon}")
        print(f"Battles won: {self.battles_won}")

    def introduce(self):
        print(f"I am {self.name} of {self.kingdom}. My weapon is a {self.weapon} and I have won {self.battles_won} battles.")

    def fight(self):
        print(f"Sir {self.name} enters the tournament with a {self.weapon}!")

    def give_advice(self):
        print(f"{self.name} recommends strengthening the kingdom's army.")
````

Notice a few things:

`Knight(RoyalMember)` means Knight inherits from `RoyalMember`.

`super().__init__()` calls the constructor of the parent class.

The Knight class receives all the attributes of RoyalMember and adds two of its own.

## Part 3: A Prince Has His Own Way of Doing Things (Method Overriding)

The King has now introduced his son, Prince Edward, to the royal court.

Like knights, princes are also members of the kingdom. However, a prince introduces himself differently.

A knight might introduce himself by mentioning his title and kingdom, while a prince might mention his royal status.

We can achieve this using **method overriding**.

Method overriding occurs when a child class provides its own implementation of a method that already exists in its parent class.

````python
class Prince(RoyalMember):
    def __init__(self, name, age, kingdom, title):
        super().__init__(name, age, kingdom)
        self.title = title

    def introduce(self):
        print(f"I am {self.title} {self.name}, heir to the throne of {self.kingdom}.")

    def give_advice(self):
        print(f"{self.title} {self.name} recommends expanding the kingdom's territory.")
````

Notice that `introduce()` already exists in `RoyalMember`. However, the `Prince` class provides its own version.

Q: What happens when a `Knight` object calls `introduce()`? What happens when a `Prince` object calls `introduce()`? Both classes inherit from `RoyalMember`. Why do they produce different outputs?

A: When a `Knight` object calls `introduce()`, it executes the overridden version defined in the `Knight` class (printing their weapon and battles won). When a `Prince` object calls `introduce()`, it executes the overridden version in the `Prince` class (printing their title and heir status). Even though both inherit from `RoyalMember`, they produce different outputs because of **method overriding**, which allows each child class to specify unique behavior for an inherited method name.

## Part 4: One Royal Court, Many Personalities (Polymorphism)

The King has finally assembled his royal court.

It includes princes and knights, all of whom have their own versions of the `introduce()` method.

Rather than calling each object's method separately, we can store all the members in one list and use a loop.

This is an example of **polymorphism**.

Polymorphism allows different objects to respond to the same method call in their own way.

````python
members = [
    RoyalMember("William", 45, "Eldoria"),
    Knight("Lancelot", 30, "Eldoria", "Sword", 12),
    Prince("Edward", 21, "Eldoria", "Prince"),
    Knight("Cedric", 27, "Northvale", "Spear", 8)
]

for member in members:
    member.introduce()

# Even though every object receives the same method call, member.introduce(),
# the output depends on the type of object.
````

### Task 5: Build the tournament system

1) Step 1: Add a method to the `Knight` class

- Create a method called `fight()` that prints:

  `Sir [name] enters the tournament with a [weapon]!`

  For example:

  `Sir Lancelot enters the tournament with a Sword!`

2) Step 2: Create the tournament participants

- Create at least four knights with different:

  - a) *Names*
  - b) *Kingdoms*
  - c) *Weapons*
  - d) *Numbers of battles won*

  Store all the knights in a list called `tournament`.

3) Step 3: Display the participants

- Use a loop to:

  - Call the `introduce()` method for every knight.
  - Call the `fight()` method for every knight.
  - Print their number of battles won.

4) Step 4: Find the most experienced knight

- Find the knight with the highest number of battles won.
- Print their name and the number of battles they have won.


````python
# STEP 1 - fight() method was added to the Knight class definition in cell 968b9f3f-9008-4934-b492-f050b73d0dc4

# STEP 2 - CREATE A LIST OF PARTICIPANTS CALLED tournament
tournament = [
    Knight("Sir Lancelot", 30, "Eldoria", "Sword", 12),
    Knight("Sir Cedric", 27, "Northvale", "Spear", 8),
    Knight("Sir Gareth", 32, "Northvale", "Axe", 15),
    Knight("Sir Galahad", 28, "Eldoria", "Mace", 10)
]

# STEP 3 - LOOP OVER THE LIST, CALL introduce(), fight() AND PRINT THE NUMBER OF BATTLES WON
for knight in tournament:
    knight.introduce()
    knight.fight()
    print(f"Battles won: {knight.battles_won}")
    print("-" * 40)

# STEP 4 - FIND THE MOST EXPERIENCED KNIGHT
most_experienced = max(tournament, key=lambda k: k.battles_won)
print(f"The most experienced knight is {most_experienced.name} with {most_experienced.battles_won} battles won!")
````

Q: How does polymorphism allow different objects to use the same method name?

A: Polymorphism allows different objects to use the same method name because Python looks up the method at runtime on the actual object type, dynamically executing the specific implementation defined by that class (dynamic dispatch).

A: One advantage of using inheritance is code reuse. It avoids duplicating standard properties and behaviors (like name, age, kingdom, and base introductions) in multiple independent classes, which makes maintenance, updates, and testing much easier.

## Final Challenge: Expand the Kingdom!

The King has decided to expand his court and recruit **Royal Guards**.

Create a new class called `RoyalGuard` that inherits from `RoyalMember`.

**Each guard should have:**
- *A name*
- *An age*
- *A kingdom*
- *A post assigned to them (e.g. Castle Gate, Throne Room, Royal Gardens)*

**Your class should:**
1. Use `super()` to initialise the inherited attributes.
2. Add a new method called `guard_duty()` that prints the guard's assigned post.
3. Override the `introduce()` method to introduce the guard and mention their assigned post.

**Now put your class to use:**
1. Create at least two `RoyalGuard` objects.
2. Add them to the existing `royal_court` list.
3. Loop through the entire list and call `introduce()` on every member.

**Final checkpoint:** Your royal court should now contain `RoyalMembers`, `Knights`, `Princes` and `RoyalGuards`, all responding to the same `introduce()` call in their own way.

````python
class RoyalGuard(RoyalMember):
    def __init__(self, name, age, kingdom, post):
        super().__init__(name, age, kingdom)
        self.post = post

    def guard_duty(self):
        print(f"{self.name} is guarding the {self.post}.")

    def introduce(self):
        print(f"I am Royal Guard {self.name} of {self.kingdom}. My post is the {self.post}.")

    def give_advice(self):
        print(f"{self.name} recommends improving the security of the castle.")

guard1 = RoyalGuard("Captain George", 35, "Eldoria", "Throne Room")
guard2 = RoyalGuard("Sentry Marcus", 29, "Eldoria", "Castle Gate")

royal_court.append(guard1)
royal_court.append(guard2)

print("--- Complete Royal Court ---")
for member in royal_court:
    member.introduce()
````

## Part 6: The King's Grand Council – The Ultimate Royal Challenge! (optional!!!)

The King has summoned all members of the kingdom for an important council meeting. Each member has different responsibilities and recommendations for improving the kingdom.

The King wants to build a **Royal Council Management System** to keep track of his council members and their advice.

### Your Tasks:

**Step 1: Create a new method**

Add a method called `give_advice()` to the `RoyalMember` class.

The method should print:

`[Name] has no official recommendation for the kingdom.`

**Step 2: Override the method**

Each type of royal member should have its own version of `give_advice()`:

- **Prince:** Recommends expanding the kingdom's territory.
- **Knight:** Recommends strengthening the kingdom's army.
- **RoyalGuard:** Recommends improving the security of the castle.

**Step 3: Assemble the council**

1. Create a list called `royal_council`.
2. Add at least one `RoyalMember`, two `Knights`, two `Princes` and two `RoyalGuards`.
3. Use a loop to make every member introduce themselves and give their advice.

**Step 4: Count the council members**

Using `isinstance()`, count how many members belong to each class.

Print the total number of:
- RoyalMembers
- Knights
- Princes
- RoyalGuards

**Step 5: The King's final decision**

The King wants to know how many council members recommended improving the kingdom's military or security.

Create a variable called `security_recommendations` and count the number of Knights and RoyalGuards in the council.

Print the final count.


````python
# Step 1 & 2: give_advice() is defined on RoyalMember, Knight, Prince, and RoyalGuard in their respective class cells.

# Step 3: Assemble the council members
royal_council = [
    RoyalMember("Arthur", 25, "Eldoria"),
    Knight("Sir Lancelot", 30, "Eldoria", "Sword", 12),
    Knight("Sir Cedric", 27, "Northvale", "Spear", 8),
    Prince("Edward", 21, "Eldoria", "Prince"),
    Prince("William", 23, "Eldoria", "Crown Prince"),
    RoyalGuard("Captain George", 35, "Eldoria", "Throne Room"),
    RoyalGuard("Sentry Marcus", 29, "Eldoria", "Castle Gate")
]

print("--- Council Meeting Advice ---")
for member in royal_council:
    member.introduce()
    member.give_advice()
    print()

# Step 4: Count the council members using isinstance() or type()
royal_member_count = 0
knight_count = 0
prince_count = 0
guard_count = 0

for member in royal_council:
    if type(member) is RoyalMember:
        royal_member_count += 1
    elif isinstance(member, Knight):
        knight_count += 1
    elif isinstance(member, Prince):
        prince_count += 1
    elif isinstance(member, RoyalGuard):
        guard_count += 1

print("--- Council Demographics ---")
print(f"Base Royal Members: {royal_member_count}")
print(f"Knights: {knight_count}")
print(f"Princes: {prince_count}")
print(f"Royal Guards: {guard_count}")

# Step 5: The King's final decision
security_recommendations = 0
for member in royal_council:
    if isinstance(member, (Knight, RoyalGuard)):
        security_recommendations += 1

print(f"\nNumber of security/military recommendations: {security_recommendations}")
````

   **Polymorphism**, which is implemented here via **method overriding** in the subclasses.

3. **What would happen if the `Knight` class did not override `give_advice()`?**
   If it did not override the method, a `Knight` instance would fall back to the parent `RoyalMember` implementation, printing that they have no official recommendation.

4. **Why is using inheritance helpful when building a system with several related classes?**
