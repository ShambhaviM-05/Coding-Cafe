## 🪄 Part 1: The Anatomy of a Spell (Functions & Control Flow)

When your Python spellbook runs, execution always begins at the first top-level statement, while any comment lines starting with `#` are completely ignored. 

* **The Main Segment:** Python internally names the top-level segment of your program as `__main__`. It stores this name in a built-in variable called `__name__`. Execution of the program begins from this `__main__` segment.
* **The `def` Keyword:** When Python sees a function definition (a `def` statement), it checks the function header to ensure it is proper, but it completely skips executing the function body until the function is actually called.
* **Control Flow:** A function call sends the control flow directly to the function definition. The spell ends when it hits a `return` statement or the last statement of the function body, which sends the control flow right back to where the function was called. 
* **Function Types:** Functions can be void (returning nothing) or non-void (returning a value), and they can be written with or without arguments. A function can also define another function inside itself, or call any other function in the same program.

## 🧪 Part 2: Potion Ingredients (Arguments & Parameters)

To pass ingredients into your spells, you use parameters and arguments. The names in the function header that catch incoming values are called **formal parameters** (or formal arguments). The actual values you send in are called **actual parameters** (or actual arguments), and they can be literals, variables, or expressions. 

You can even use a function call inside another statement (like `print()`) or as part of an expression. Just keep in mind that the mutability or immutability of these arguments plays an important role when passing values.

**The Strict Rules of Syntax:**
* **Positional First:** In a function call statement, positional (required) arguments must always come *before* any keyword arguments.
* **No Double Dipping:** You cannot specify a value for an argument more than once.
* **Keyword Arguments:** As long as you name the arguments, you can write them in any order you please.
* **Default Arguments:** You can assign a default value to a parameter in the function header, so if a value is skipped during the call, Python fills it in.

*Crucial Syntax Rule:* **Non-default arguments cannot follow a default argument** in the header.

````python
# 'player_score' is a required positional argument.
# 'xp_boost' has a default value of 10!
def add_points(player_score, xp_boost=10):
    total = player_score + xp_boost
    print("your total points are :", total)

# answer here
# 1) you only pass the score. The bonus defaults to 10.
  
# 2) you pass both! The default 10 is replaced by some other number.
````

Python Argument Rules: Player Score Edition : When calculating a player's score in Python, your argument list must first contain positional (required) arguments followed by any keyword arguments. Having a positional argument after keyword arguments will result into an error. Furthermore, you cannot specify a value for an argument more than once.  

The Core Function:

````python
def calculate_score(score, level, bonus=2, multiplier=0.05):
    return score * level * bonus * multiplier
````

| Function Call Statement | Status | System Reason |
|---|---|---|
| `calculate_score(score=3000, level=5)` | Legal | Non-default values provided as named arguments. |
| `calculate_score(multiplier=0.12, score=5000, level=4)` | Legal | Keyword arguments can be used in any order, and the skipped argument has a default value. |
| `calculate_score(level=4, multiplier=0.12, score=5000)` | Legal | With keyword arguments, we can give values in any order. |
| `calculate_score(5000, 3, multiplier=0.05)` | Legal | Positional arguments come before keyword arguments; the skipped argument has a default value. |
| `calculate_score(multiplier=0.05, 5000, 3)` | Illegal | Keyword argument before positional arguments. |
| `calculate_score(5000, score=300, level=2)` | Illegal | Multiple values provided for `score`: once as a positional argument and again as a keyword argument. |
| `calculate_score(5000, points=300, level=2)` | Illegal | Undefined name used (`points` is not a parameter). |
| `calculate_score(500, bonus=2, multiplier=0.05)` | Illegal | A required argument (`level`) is missing. |

## 🔮 Part 3: The Crystal Ball of Visibility (Scope & The LEGB Rule)

Not all variables can be seen from everywhere! 
* **Global Scope:** A name declared in the top-level segment (`__main__`) has a global scope, meaning it is usable inside the whole program and all blocks within it.
* **Local Scope:** A name declared inside a function body (including formal arguments) has local scope and can only be used within that function.

When a function is called, a local environment is created for it; if it calls another function, a nested local environment is created. 

**The LEGB Resolution Rule:**
Whenever Python looks for a variable name, it checks environments in this exact order:
1. **L**ocal environment
2. **E**nclosing environment
3. **G**lobal environment
4. **B**uilt-in environment 

*Important Trap:* Whenever you assign a value to a variable inside a function, Python automatically creates a local variable. If you actually want to assign a value to a *global* variable from inside a function, you must use the `global` statement.

## 🏛️ Part 4: The Grand Library (Modules)

Partitioning a program into individual components is called **modularity**; it reduces complexity and allows you to reuse contents in other programs without rewriting them. 
* A **Python module** is simply a `.py` file containing docstrings, variables, classes, objects, statements, and functions. 
* A **library** is a collection of modules (and packages) that cater to a specific type of application, such as the Python standard library or the NumPy library.

**Importing Syntax:**
* `import <module>`: This executes the module's code and sets up a new namespace. You **must** prefix imported objects with the module name (e.g., `myMod.checknum()`).
* `from <module> import <object>`: This brings specific objects directly into your environment. You **must not** qualify the imported item with the module name (e.g., use `pi`, not `math.pi`).

````python
# example code
import time
print("Casting spell in 3...")
time.sleep(1)  # Pauses the program for 1 second
print("2...")
time.sleep(1)
print("1...")
time.sleep(1)

print("RIBBIT! You are now a frog.")
````

````python
# example code
import random

print(" Boss defeated! Generating loot... \n")

# 1. randint(a, b): Gives an exact integer (both limits inclusive)
coins = random.randint(50, 100)
print(f" Gold coins found: {coins}")

# 2. random(): Gives a decimal between 0.0 and 1.0
crit_chance = random.random()
print(f" Critical hit chance upgrade: {round(crit_chance, 2)}")

# 3. uniform(a, b): Gives a decimal between a and b
potion_power = random.uniform(1.5, 3.5)
print(f" Potion multiplier: {round(potion_power, 2)}x")

# 4. randrange(start, stop, step): Picks from 10, 20, 30, or 40!
bonus_xp = random.randrange(10, 50, 10)
print(f" Bonus XP gained: {bonus_xp}")
````

## 📜 Part 5: Scribing Scrolls (Data & File Handling)
To preserve your data after your program finishes, you must write it to a text file (a scroll). 

**Essential File Syntax:**
Use the `with open()` statement to ensure your scroll rolls itself back up (closes) automatically!
* **Read Mode (`'r'`):** `with open('scroll.txt', 'r') as file:` lets you read the contents without altering them. Use `file.read()` to get the whole text or `file.readlines()` to get a list of lines.
    - we haven't covered lists yet, for now just know that it is a collection of values/elements/variables and you can loop over it like you did with strings
* **Write Mode (`'w'`):** `with open('scroll.txt', 'w') as file:` creates a brand new scroll or completely erases an existing one to write fresh data using `file.write()`.
* **Append Mode (`'a'`):** `with open('scroll.txt', 'a') as file:` lets you safely add new ink to the very bottom of an existing scroll without destroying what was written before.

````python
# example code
ingredients = ["broccoli", "carrots", "garlic", "ginger"]

# 1. Writing the recipe scroll (creates or overwrites the file)
with open("goodsoup.txt", "w") as scroll:
    for item in ingredients:
        scroll.write(f"Chop and add {item}\n")

# 2. Reading it back to make sure the ink dried properly!
with open("goodsoup.txt", "r") as scroll:
    print("The ancient recipe says:\n")
    print(scroll.read())
````

A wandering ranger must track their daily wellness habits to keep their stamina up for long journeys into the mountains.
Write a program that asks the user to input their daily step count and their daily fluid intake in liters. Create a function that evaluates these inputs. If the goals are not met, append *"Rest and Hydrate at the Tavern."* Ensure the file is opened in append mode ('a') so previous days' records are not overwritten.


````python
def evaluate_stamina(steps, fluids):
  
    if steps > 10000 and fluids > 3:
        message = "Quest Ready for the Triund Trek!"
    else:
        message = "Rest and Hydrate at the Tavern."

    with open("ranger_stamina_log.txt", "a") as file:
        file.write(message + "\n")

    return message

if __name__ == "__main__":
    steps = int(input("Enter your daily step count: "))
    fluids = float(input("Enter your fluid intake in liters: "))

    result = evaluate_stamina(steps, fluids)
    print(result)
````

You are the Dungeon Master. Your task is to write a simulation for heroes attempting the dangerous **"Trial of Shadows."** The simulation must build suspense, generate a random score, save the hero's fate to a permanent log, and return the final result!

Instructions:

````python
import time
import random

def attempt_trial(hero_name):
    print(f"{hero_name} enters the Trial of Shadows...")

  
    time.sleep(2)

    score = random.randint(1, 100)

    if score >= 50:
        outcome = "survived"
        print(f"{hero_name} scored {score} and SURVIVES the trial!")
    else:
        outcome = "failed"
        print(f"{hero_name} scored {score} and FAILS the trial...")

    with open("hall_of_fame.txt", "a") as file:
        file.write(f"{hero_name} scored {score} and {outcome} the Trial of Shadows.\n")

    return score

if __name__ == "__main__":
    final_score = attempt_trial("Aria")
    print(f"GAME OVER! Final score: {final_score}")
````
